```python
print("==Sharmeen's Travel Agency==")
from_city = input("\nEnter your Departure: ")
to_city = input("Enter your Destination: ")

passengers = int(input("Enter number of passengers: "))

if from_city == "Karachi" and to_city == "Lahore":

    print("\n1. Flight 101 - Rs.15000")
    print("2. Flight 102 - Rs.18000")

    choice = int(input("Choose Flight: "))

    if choice == 101:
        price = 15000

    elif choice == 102:
        price = 18000

    print("\nTotal price: Rs.", price * passengers)
    print("Flight booked successfully!")

else:
    print("No flights available.")
```
